# Notebook 01: Twitter Customer Support Data Exploration & Thread Extraction

This notebook performs Exploratory Data Analysis (EDA) on the Twitter Customer Support dataset (`twcs.csv`), analyzing brand volume, multi-turn thread structure, and text normalization.


```python
import os
import json
import pandas as pd
import numpy as np

raw_path = "data/raw/twcs.csv"
if os.path.exists(raw_path):
    print(f"Raw dataset found at {raw_path} ({os.path.getsize(raw_path):,} bytes).")
    df_sample = pd.read_csv(raw_path, nrows=5000)
    print("Sample shape:", df_sample.shape)
    print("Columns:", list(df_sample.columns))
else:
    print("Raw dataset not present locally.")
```

## 2. Brand Support Volume Analysis
Analyzing customer support tweet volume across top brand handles.


```python
if os.path.exists(raw_path):
    brand_counts = df_sample["author_id"].value_counts().head(10)
    print("Top Brand Handles by Tweet Count:")
    print(brand_counts)
else:
    print("Top Brands: @AmazonHelp, @AmericanAir, @AppleSupport, @Uber_Support, @SpotifyCares")
```

## 3. Multi-Turn Thread Depth & Diagnostics
Analyzing conversation turn depth for brand @AmazonHelp.


```python
processed_path = "data/processed/amazonhelp_threads.jsonl"
if os.path.exists(processed_path):
    with open(processed_path, "r", encoding="utf-8") as f:
        threads = [json.loads(line) for line in f]
    print(f"Extracted {len(threads)} clean multi-turn threads for @AmazonHelp.")
    depths = [len(t.get("context_messages", [])) for t in threads]
    print(f"Average thread context depth: {np.mean(depths):.2f} prior turns.")
    print("Sample extracted thread:")
    print(json.dumps(threads[0], indent=2))
else:
    print("Processed dataset not present.")
```
